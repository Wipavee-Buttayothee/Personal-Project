What each column represent?
* ticket_id: the report's unique ID
* timestamp: when it was submitted (likely in UTC, so add 7 hours for Bangkok time)
* province, district, subdistrict: address
* type: the problem category, sometimes several (e.g. "flooding, roads")
* comment: the citizen's description in Thai
* is_... (25 columns): the problem category, with 1 if the report belongs to that category and 0 if not (e.g. is_น้ำท่วม for flooding)

```python
# Download the list of all Thai districts with English names
url = "https://raw.githubusercontent.com/kongvut/thai-province-data/master/api/latest/district.json"
districts = pd.read_json(url)

# Bangkok districts in this list start with "เขต" / "Khet", so remove that to match your data
districts["name_th"] = districts["name_th"].str.replace("^เขต", "", regex=True).str.strip()
districts["name_en"] = districts["name_en"].str.replace("^Khet ", "", regex=True).str.strip()

district_lookup = dict(zip(districts["name_th"], districts["name_en"]))
district_lookup["ไม่ระบุ"] = "Unspecified"

# Translate every district, in Bangkok and other provinces
df["district_en"] = df["district"].map(district_lookup)

# Check: any districts that didn't get translated?
print(df.loc[df["district_en"].isna(), "district"].value_counts())
```

```python
# 4. Correct time
df["timestamp"] = df["timestamp"] + pd.Timedelta(hours=7)

# 5. Remove duplicates
df["date"] = df["timestamp"].dt.date
before = len(df)
df = df.drop_duplicates(subset="ticket_id")
df = df.drop_duplicates(subset=["comment", "district", "date"])
print("Removed duplicates:", before - len(df))
```

```python
# 6. columns for analysis
df["comment"] = df["comment"].fillna("")
df["month"] = df["timestamp"].dt.to_period("M").dt.to_timestamp()
df["year"] = df["timestamp"].dt.year
df["month_of_year"] = df["timestamp"].dt.month
df["hour"] = df["timestamp"].dt.hour
df["weekday"] = df["timestamp"].dt.day_name()

is_cols = [c for c in df.columns if c.startswith("is_")]
df["n_categories"] = df[[c for c in is_cols if c != "is_ไม่ระบุ"]].sum(axis=1)
df["informative_text"] = df["comment"].str.len() >= 10
```
